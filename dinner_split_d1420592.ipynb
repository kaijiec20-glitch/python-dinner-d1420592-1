```python
import re  # 匯入正規表達式模組，用來從雜亂的文字中精準抓出數字

def calculate_dinner_cost(raw_cost_str, raw_people_str, discount_coupon, budget_per_person=300):
    """
    聚餐費用試算器函式
    
    參數說明：
    raw_cost_str (str): 餐廳提供的原始餐費文字（例如 "$3,500", "NT$ 8,800"）
    raw_people_str (str): 聚餐人數文字（例如 "5 人", "共 6 位"）
    discount_coupon (int/float): 折價券折抵金額
    budget_per_person (int/float): 每人預算上限，題目規定預設為 300 元
    """
    
    # ==========================================
    # 步驟 1：資料清洗與型別轉換
    # ==========================================
    
    # 使用 re.findall(r"\d+", ...) 找出字串中所有的連續數字（會回傳一個字串清單，例如 ['3', '500']）
    # 再用 "".join(...) 把清單裡的數字無縫接起來，變成乾淨的純數字字串（例如 "3500"）
    cost_digits = "".join(re.findall(r"\d+", raw_cost_str))
    people_digits = "".join(re.findall(r"\d+", raw_people_str))

    # 將整理好的純數字字串，轉換成整數 (int) 型別，這樣才能進行加減乘除運算
    original_cost = int(cost_digits)
    people_count = int(people_digits)

    # ==========================================
    # 步驟 2：依照聚餐規則進行費用計算
    # ==========================================
    
    # 規則 1：餐廳按「原始餐費」收取 10% 服務費
    service_fee = original_cost * 0.10
    
    # 計算加上服務費後的金額
    cost_with_fee = original_cost + service_fee
    
    # 規則 2：計算服務費後，再扣除折價券
    # 使用 max(0.0, ...) 確保即使折價券金額大於餐費，總金額最低只會是 0 元，不會變成負數
    final_total_cost = max(0.0, cost_with_fee - discount_coupon)
    
    # 規則 3：所有人平均分攤應付總額
    # 加上 if people_count > 0 是為了防止人數為 0 時發生「除以零 (ZeroDivisionError)」的程式錯誤
    per_person_cost = final_total_cost / people_count if people_count > 0 else 0
    
    # 規則 4：判斷每人分攤費用是否超過 300 元預算
    # 比較結果會是布林值：True（超過預算）或 False（未超過預算）
    is_over_budget = per_person_cost > budget_per_person
    
    # 計算每人費用與預算的差額，使用 abs() 取絕對值，方便顯示「多付多少」或「省下多少」
    budget_diff = abs(per_person_cost - budget_per_person)

    # ==========================================
    # 步驟 3：產生容易閱讀的費用摘要報告
    # ==========================================
    
    print("=" * 42)  # 印出 42 個等號作為報表上邊框
    print("🍽️  聚餐費用試算摘要報告")
    print("=" * 42)
    
    # 使用 f-string 格式化輸出，{數值:,} 可以自動幫數字加上千分位逗號（例如 3500 -> 3,500）
    print(f"• 原始餐費文字 : {raw_cost_str}  👉 轉換數值：{original_cost:,} 元")
    print(f"• 參與人數文字 : {raw_people_str}  👉 轉換人數：{people_count} 人")
    print(f"• 使用折價券   : {discount_coupon:,} 元")
    print("-" * 42)  # 印出分隔線
    
    # :,.1f 代表顯示千分位逗號，並四捨五入到小數點後第 1 位
    print(f"• 10% 服務費   : {service_fee:,.1f} 元")
    print(f"• 最終應付總額 : {final_total_cost:,.1f} 元")
    
    # :,.2f 代表顯示千分位逗號，並四捨五入到小數點後第 2 位
    print(f"• 每人平均分攤 : {per_person_cost:,.2f} 元")
    print(f"• 每人預算限制 : {budget_per_person} 元")
    
    # 根據前面的布林值判斷結果，印出對應的提示訊息
    if is_over_budget:
        print(f"⚠️  預算結果：超過預算！每人需多付 {budget_diff:,.2f} 元")
    else:
        print(f"✅  預算結果：符合預算！每人省下了 {budget_diff:,.2f} 元")
        
    print("=" * 42 + "\n")  # 印出下邊框並空一行（\n），讓多組報表之間有間距


# ==========================================
# 主程式區塊：執行三組測試資料
# ==========================================
# 當這個 Python 檔案被直接執行時，才會執行以下縮排內的測試程式碼
if __name__ == "__main__":
    print("開始執行三組聚餐費用測試...\n")
    
    # 【測試組 1】：一般情況，有貨幣符號與逗號，結果超過預算
    calculate_dinner_cost(
        raw_cost_str="$1,800", 
        raw_people_str="5 人", 
        discount_coupon=100
    )
    
    # 【測試組 2】：較大金額且帶有 NT$ 前綴，人數文字帶有中文前綴，結果超過預算
    calculate_dinner_cost(
        raw_cost_str="NT$ 4,500", 
        raw_people_str="共 12 位", 
        discount_coupon=500
    )
    
    # 【測試組 3】：餐費文字後面有「元」字，折價後剛好在預算內
    calculate_dinner_cost(
        raw_cost_str="1,200 元", 
        raw_people_str="4個人", 
        discount_coupon=200
    )
```
